Polynomial regression example using a small synthetic salary-position dataset.

```python
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
plt.rcParams['figure.figsize'] = (7, 4)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error

rng = np.random.default_rng(RANDOM_STATE)
experience = np.arange(1, 16).reshape(-1, 1)
salary = 28000 + 3500*experience.ravel() + 550*(experience.ravel()**2) + rng.normal(0, 4000, len(experience))
dataset = pd.DataFrame({'ExperienceYears': experience.ravel(), 'Salary': salary.astype(int)})
dataset
```

```python
lin = LinearRegression().fit(experience, salary)
poly = make_pipeline(PolynomialFeatures(degree=2), LinearRegression()).fit(experience, salary)
grid = np.linspace(1, 15, 200).reshape(-1, 1)

print('Linear RMSE:', np.sqrt(mean_squared_error(salary, lin.predict(experience))))
print('Polynomial RMSE:', np.sqrt(mean_squared_error(salary, poly.predict(experience))))

plt.figure()
plt.scatter(experience, salary, label='data')
plt.plot(grid, lin.predict(grid), label='linear')
plt.plot(grid, poly.predict(grid), label='polynomial degree 2')
plt.xlabel('Experience years'); plt.ylabel('Salary')
plt.legend(); plt.grid(True); plt.show()
```
